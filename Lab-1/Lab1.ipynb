N Queens Problem


```python
from pprint import pprint
```

```python
def is_safe(board, x, y, n):
    # Column Test
    for row in range(x):
        if board[row][y] == 'Q':
            return False

    # Left Upper Diagonal Test
    r = x - 1
    c = y - 1
    while(r>=0 and c>=0):
        if(board[r][c] == 'Q'):
            return False
        r = r - 1
        c = c - 1

    #Right Upper Diagonal Test
    r = x - 1
    c = y + 1
    while(r>=0 and c<n):
        if(board[r][c] == 'Q'):
            return False
        r = r - 1
        c = c + 1
    # Return True if all the Tests are passed
    return True
```

```python
n = int(input("Enter number of Q's"))
board = [['_']*n for i in range(n)]
def nQueenSolver(board, row, n):
    if(row>=n):
        return True
    for col in range(n):
        if is_safe(board,row,col,n):
            board[row][col] = 'Q'
            if(nQueenSolver(board, row+1,n)):
                return True
            board[row][col] = '_'
    return False

if(nQueenSolver(board,0,n)):
    pprint(board)
else:
    pprint("NO SOLUTION")
```

Task 1: Goat, Wolf and Cabbage Problem

```python
from collections import deque

def is_safe(state):
    F, G, W, C = state

    if W == G and F != W:
        return False
    if G == C and F != G:
        return False
    
    return True

def get_next_states(state):
    F, G, W, C = state 
    next_states = []

    # Farmer crosses alone
    new_state = (1-F, G, W, C)
    if is_safe(new_state):
        next_states.append((new_state, "Farmer crosses alone"))

    # Farmer takes Goat
    if F == G:
        new_state = (1-F, 1-G, W, C)
        if is_safe(new_state):
            next_states.append((new_state, "Farmer takes Goat"))

    # Farmer takes Wolf
    if F == W:
        new_state = (1-F, G, 1-W, C)
        if is_safe(new_state):
            next_states.append((new_state, "Farmer takes Wolf"))

    # Farmer takes Cabbage
    if F == C:
        new_state = (1-F, G, W, 1-C)
        if is_safe(new_state):
            next_states.append((new_state, "Farmer takes Cabbage"))  

    return next_states
 

def bfs():
    start = (0, 0, 0, 0)
    goal = (1, 1, 1, 1)

    queue = deque([(start, [])])
    visited = set()

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path
        if state in visited:
            continue
        visited.add(state)

        for next_state, action in get_next_states(state):
            queue.append((next_state, path + [action]))

    return None

solution = bfs()

print("Solution:") 
for i, step in enumerate(solution, start=1):
    print(f"{i}. {step}")
```

Task 2: Water Jug Problem

```python
from collections import deque

def get_next_states(state):
    A, B = state
    next_states = []
    
    next_states.append(((4,B), "Fill Jug A"))
    next_states.append(((A,3), "Fill Jug B"))
    next_states.append(((0,B), "Empty  Jug A"))
    next_states.append(((A,0), "Empty Jug B"))

    # Pour A->B
    transfer = min(A, 3-B)
    next_states.append(
        ((A-transfer, B+transfer),
         "Pour Jug A -> Jug B")
    ) 

    # Pour B->A
    transfer = min(B, 4-A)
    next_states.append(
        ((A+transfer, B-transfer),
         "Pour Jug B -> Jug A")
    )

    return next_states

def bfs():
    start = (0,0) 

    queue = deque([(start, [start])])
    visited = set()

    while queue:
        state, path = queue.popleft()

        A,B = state

        # Goal Test
        if A == 2:
            return path
        
        if state in visited:
            continue

        visited.add(state)

        for next_state, action in get_next_states(state):
            if next_state not in visited:
                queue.append(
                    (next_state, path + [next_state])
                )

    return None
    
solution = bfs()

print("shortest solution path:")
for state in solution:
    print(state)
```

Task 3: Robot Navigation Problem

```python
from collections import deque

grid = [
    ['S', '.', '.', 'X', '.'],
    ['.', 'X', '.', 'X', '.'],
    ['.', 'X', '.', '.', '.'],
    ['.', '.', 'X', 'X', '.'],
    ['.', '.', '.', '.', 'G']
]

ROWS = 5
COLS = 5

start = (0, 0)
goal = (4, 4)

# Up, Down, Left, Right
directions = [
    (-1, 0),
    (1, 0),
    (0, -1),
    (0, 1)
]


def bfs():
    queue = deque([(start, [start])])
    visited = set()

    while queue:
        current, path = queue.popleft()

        if current == goal:
            return path

        if current in visited:
            continue

        visited.add(current)

        r, c = current

        for dr, dc in directions:
            nr = r + dr
            nc = c + dc

            if (0 <= nr < ROWS and 0 <= nc < COLS and grid[nr][nc] != 'X' and (nr, nc) not in visited):
                queue.append(
                    ((nr, nc), path + [(nr, nc)])
                )

    return None

path = bfs()

print("Shortest Path:")
for cell in path:
    print(cell)
```
