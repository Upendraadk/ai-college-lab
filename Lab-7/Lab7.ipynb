##  Multimodal Pipeline:  Image Captioning, Translation, and Text-to-Speech Objective

### Part A: Image Captioning

```python
from huggingface_hub import login
login(token='Use your Hugging Face token here')
```

```python
# Import Required Libraries
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import torch
```

```python
# Load the processor and model
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")
```

```python
# Load an Image
image = Image.open("images.jpg").convert('RGB')
```

```python
# Process the image
inputs = processor(image, return_tensors="pt")
```

```python
# Generate caption
with torch.no_grad():
    output = model.generate(**inputs)
```

```python
# Decode and print the caption
caption = processor.decode(output[0], skip_special_tokens=True)
print("Caption:", caption)
```

### Part B: Text Translation

```python
# Translate English Text to Nepali
from googletrans import Translator

translator = Translator()
result = await translator.translate("How are you?", src='en', dest='ne')
print(result.text)
```

### Part C: Text-to-Speech

```python
# Convert Nepali Text to Speech and Save/Play It
from gtts import gTTS
import os
# Nepali text
text = "  तपाईंलाई कस्तो छ?"
# Convert text to speech
tts = gTTS(text=text, lang='ne')
# Save as mp3
tts.save("nepali_speech.mp3")
# Play the audio 
from IPython.display import Audio
Audio("nepali_speech.mp3", autoplay=True)
```
